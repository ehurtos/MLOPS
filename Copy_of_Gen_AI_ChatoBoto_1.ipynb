```python
#Create a endless loop gor a chatbot using Google AI model and setting tone of annwers for model to respnd.

import google.generativeai as genai

# Configure the API key (replace with your own key)
genai.configure(api_key="AIz")

# Set up the model
model = genai.GenerativeModel("gemini-1.5-flash")

# Initialize the messages array with a system message
messages = [
    {
        "role": "system",
        "content": "You are an AI assistant named Edusko. You always provide answers so 12 year old kids would undestand."
    }
]

while True:
    # Get user input
    user_message = input("User: ")

    if user_message:
        # Append user message to the messages array
        messages.append({"role": "user", "content": user_message})

        # Generate a response using the Google Generative AI API
        response = model.generate_content(messages[-1]["content"])

        # Print the generated response
        print("Claude:", response.text)

        # Append the assistant's response to the messages array
        messages.append({"role": "assistant", "content": response.text})
```
